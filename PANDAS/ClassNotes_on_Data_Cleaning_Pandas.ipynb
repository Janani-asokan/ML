## 1. What is Data Cleaning?

**Definition:** Data Cleaning is the process of identifying and correcting inaccurate,
incomplete, duplicate, inconsistent, or improperly formatted data to improve data quality
before analysis.

**Goal:** Convert ❌ Raw Data → ✅ High-Quality Data

**The Data Cleaning Process (pipeline):**

`Raw Data (errors, missing values, duplicates)` → `Cleaning (identify & fix, standardize)` →
`Reliable Data (clean, consistent, accurate, complete)` → `Analysis (better insights & decisions)`

### Why is Data Cleaning Important?
- **Improves Accuracy** — accurate data gives accurate insights
- **Removes Errors** — fixes mistakes and inconsistencies
- **Reduces Noise** — keeps only relevant and useful data
- **Better Machine Learning Models** — clean data improves model performance
- **Reliable Business Decisions** — strong data leads to confident decisions
- **Saves Time during Analysis** — clean data is easy to work with

### Famous Principle
> **Garbage In → Garbage Out (GIGO)** — if the input data is bad, the output will be bad,
> no matter how good your analysis technique is.

### Common Data Quality Problems
| Problem | Example |
|---|---|
| Missing Values | Age is blank |
| Duplicate Records | Same student entered twice |
| Incorrect Entries | Name typed as `three` instead of `3` |
| Typographical Errors | `'Delhi` with a stray apostrophe |
| Invalid Data | Age = `-8` or `250` |
| Inconsistent Formatting | `delhi` vs `Delhi` vs `DELHI` |
| Extra Spaces | `" Pune "` |
| Wrong Data Types | Salary stored as text `"fifty five thousand"` |
| Outliers | Salary = 1,000,000 when everyone else earns ~50,000 |

## 12. Checking Missing Values — `df.isnull()` and `df.isnull().sum()`

**What it is:**
- `df.isnull()` returns a DataFrame of the **same shape**, with `True` where a value is
  missing and `False` where a value is present.
- `df.isnull().sum()` adds those `True`/`False` values up **column-by-column**
  (`True` counts as 1), giving you the **count of missing values per column**.

**Why we use it:** To find out exactly *which* columns have missing data and *how many* values
are missing, so you can decide how to handle them (drop, fill with mean/median/mode, etc.).


````python
df.isnull()          # True = missing, False = present (whole grid)
````

````python
df.isnull().sum()    # missing value count PER COLUMN — the most commonly used form
````

**Interpretation:** A `0` next to a column means that column has **no missing values**.
Any number greater than `0` tells you exactly how many records need attention in that column.


## 13. Detecting Duplicate Records — `df.duplicated()`

**What it is:** Returns `True` for every row that is an **exact duplicate** of a previous row
(by default, Pandas compares **all columns** to decide if a row is a duplicate).

**Why we use it:** Duplicate records inflate counts, distort averages/statistics, and can
seriously bias analysis and ML models if not removed.


````python
df.duplicated()           # True marks a row that repeats an earlier row
````

````python
df.duplicated().sum()     # total count of duplicate rows in the dataset
````

````python
# See the actual duplicate row(s)
df[df.duplicated()]
````

## 14. Removing Duplicates — `df.drop_duplicates()`

**What it is:** Returns a **new** DataFrame with duplicate rows removed (the original `df` is
left unchanged unless you pass `inplace=True` or reassign it).

**Why we use it:** Once duplicates are detected, this is the standard way to remove them and
keep only one copy of each unique record.

```python
df_clean = df.drop_duplicates()          # returns a new DataFrame
df.drop_duplicates(inplace=True)         # OR modifies df directly
```


````python
df_no_duplicates = df.drop_duplicates()
print("Shape before:", df.shape)
print("Shape after: ", df_no_duplicates.shape)
df_no_duplicates
````

## 15. Identifying Invalid Values (Logical Validation)

**What it is:** Checking values against real-world logical rules — not every error shows up
as a missing value or a duplicate. A number can be **present, correctly typed, and still
wrong** (e.g. Age = 200).

**Why we use it:** Some errors are only caught through domain knowledge / business rules, e.g.:
- Age should logically be between **0 and 120**
- Salary cannot be **negative**
- Marks cannot **exceed 100**

`df.describe()` (Section 10) is a great first pass for spotting these via min/max — but we can
also filter explicitly to see the offending rows.


````python
# Rows with an illogical Age (negative, or above a realistic human age limit)
invalid_age = df[(df["Age"] < 0) | (df["Age"] > 120)]
invalid_age
````

````python
# Rows with an illogical Marks value (assuming marks are out of 100)
invalid_marks = df[(df["Marks"] < 0) | (df["Marks"] > 100)]
invalid_marks
````

## 16. Putting It All Together — The Cleaning Workflow

The lecture summarizes the process as a pipeline:

```
Collect Dataset → Load into Pandas → Inspect Dataset → Find Missing Values
   → Detect Duplicates → Check Data Types → Identify Invalid Data
   → Clean Dataset → Ready for Analysis
```

Let's run the full inspection workflow on our dataset in one place — this is exactly what you
would type at the very start of any real data cleaning project.


````python
# STEP 1-2: Load into Pandas
df = pd.read_csv("students_data_cleaning_practice.csv")

# STEP 3: Inspect Dataset
print("First 5 rows:")
display(df.head())

print("\nShape (rows, columns):", df.shape)

print("\nColumn names:", df.columns.tolist())

print("\nInfo:")
df.info()
````

````python
# STEP 4: Find Missing Values
print("Missing values per column:")
print(df.isnull().sum())
````

````python
# STEP 5: Detect Duplicates
print("Number of duplicate rows:", df.duplicated().sum())
display(df[df.duplicated()])
````

````python
# STEP 6: Check Data Types
print(df.dtypes)
````

````python
# STEP 7: Identify Invalid Data (logical validation)
print("Invalid ages (outside 0-120):")
display(df[(df["Age"] < 0) | (df["Age"] > 120)])

print("Invalid marks (outside 0-100):")
display(df[(df["Marks"] < 0) | (df["Marks"] > 100)])
````

````python
# STEP 8: Clean the Dataset (example fixes — combine everything learned above)
df_clean = df.copy()

# a) remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

# b) fill missing City with 'Unknown' and missing Marks with the column mean
df_clean["City"] = df_clean["City"].fillna("Unknown")
df_clean["Marks"] = df_clean["Marks"].fillna(df_clean["Marks"].mean())

# c) fix invalid ages/marks by clipping to a realistic valid range
df_clean["Age"] = df_clean["Age"].clip(lower=0, upper=120)
df_clean["Marks"] = df_clean["Marks"].clip(lower=0, upper=100)

df_clean
````

````python
# STEP 9: Ready for Analysis — verify the dataset is now clean
print("Missing values left:", df_clean.isnull().sum().sum())
print("Duplicate rows left:", df_clean.duplicated().sum())
print("Any invalid ages left:", ((df_clean['Age'] < 0) | (df_clean['Age'] > 120)).sum())
print("Any invalid marks left:", ((df_clean['Marks'] < 0) | (df_clean['Marks'] > 100)).sum())
````
